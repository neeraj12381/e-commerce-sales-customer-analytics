```python
# Load the Excel sheets for MySQL data import

excel_file = "Ecommerce_Data_Analyst_Project_Dataset.xlsx"

customers = pd.read_excel(excel_file, sheet_name="Customers")
products = pd.read_excel(excel_file, sheet_name="Products")
orders = pd.read_excel(excel_file, sheet_name="Orders_Raw")

# Check the number of records in each sheet
print("Customers:", len(customers))
print("Products:", len(products))
print("Orders:", len(orders))
```

```python
# Load customer data from Excel into the MySQL Customers table

cursor = conn.cursor()

customer_query = """
INSERT INTO Customers
(Customer_ID, Customer_Name, Gender, Age, City, State)
VALUES (%s, %s, %s, %s, %s, %s)
"""

customer_data = list(
    customers[
        ["Customer_ID", "Customer_Name", "Gender", "Age", "City", "State"]
    ].itertuples(index=False, name=None)
)

cursor.executemany(customer_query, customer_data)
conn.commit()

# Check the number of inserted customer records
print("Customers inserted:", cursor.rowcount)
```

```python
# Load product data from Excel into the MySQL Products table

product_query = """
INSERT INTO Products
(Product_ID, Product, Category, Sub_Category, Cost_Price)
VALUES (%s, %s, %s, %s, %s)
"""

product_data = list(
    products[
        ["Product_ID", "Product", "Category", "Sub_Category", "Cost_Price"]
    ].itertuples(index=False, name=None)
)

cursor.executemany(product_query, product_data)
conn.commit()

# Check the number of inserted product records
print("Products inserted:", cursor.rowcount)
```

```python
# Load cleaned order data into the MySQL Orders table

orders_sql = orders_clean[
    ["Order_ID", "Order_Date", "Customer_ID", "Product_ID",
     "Quantity", "Discount", "Sales", "Profit",
     "Payment_Mode", "Shipping_Mode"]
].copy()

# Convert missing values to None for MySQL
orders_sql = orders_sql.where(pd.notnull(orders_sql), None)

order_query = """
INSERT INTO Orders
(Order_ID, Order_Date, Customer_ID, Product_ID, Quantity,
 Discount, Sales, Profit, Payment_Mode, Shipping_Mode)
VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s)
"""

order_data = list(
    orders_sql.itertuples(index=False, name=None)
)

cursor.executemany(order_query, order_data)
conn.commit()

# Check the number of inserted order records
print("Orders inserted:", cursor.rowcount)
```
